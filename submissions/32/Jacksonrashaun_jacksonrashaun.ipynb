```r
# ============================================================
# Task 2.1: Check for Missing Values
# ============================================================

# Read the dataset directly (if not already loaded)
sales_data <- read_csv("/workspaces/Fall2025-MS3083-Base_Template/data/sales_data.csv")

# TODO: Create 'missing_summary' that shows count of NAs in each column of sales_data
missing_summary <- sapply(sales_data, function(x) sum(is.na(x)))

cat("========== MISSING VALUES SUMMARY ==========\n")
print(missing_summary)
cat("\nTotal missing values:", sum(missing_summary), "\n")
```

```r
# ============================================================
# Task 2.3: Detect Outliers in Revenue
# ============================================================

# TODO: Calculate outlier thresholds using IQR method
#   - Calculate Q1 (25th percentile) and Q3 (75th percentile) of Revenue
#   - Calculate IQR = Q3 - Q1
#   - Lower bound = Q1 - 1.5 * IQR
#   - Upper bound = Q3 + 1.5 * IQR
# TODO: Create 'outlier_analysis' dataframe with these values

# --- Your code here ---
Q1 <- quantile(sales_clean$Revenue, 0.25, na.rm = TRUE)
Q3 <- quantile(sales_clean$Revenue, 0.75, na.rm = TRUE)
IQR_value <- Q3 - Q1
lower_bound <- Q1 - 1.5 * IQR_value
upper_bound <- Q3 + 1.5 * IQR_value
# ------------------------------------------------------------

outlier_analysis <- data.frame(
  Metric = c("Q1", "Q3", "IQR", "Lower Bound", "Upper Bound"),
  Value = c(Q1, Q3, IQR_value, lower_bound, upper_bound)
)

cat("========== OUTLIER ANALYSIS ==========\n")
print(outlier_analysis)

# Count outliers
outlier_count <- sum(sales_clean$Revenue < lower_bound | sales_clean$Revenue > upper_bound)
cat("\nNumber of outliers detected:", outlier_count, "\n")
```

```r
# ==========================================
# Task 6.1: Join Customers and Orders (flexible auto-locate)
# ==========================================

suppressPackageStartupMessages({
  library(dplyr)
})

# ---- Debug helpers (optional but handy) ------------------------------------
say <- function(...) cat(paste0(..., "\n"))

# ---- Smart file finder -----------------------------------------------------
# Finds the "best" match for a data file based on tokens (e.g., c("customer", "customers"))
# across CSV/XLSX, allowing prefixes/suffixes like *_clean, -2024, etc.
find_data_file <- function(tokens,
                           roots = c(
                             ".", "data", "data/raw", "data/clean", "datasets",
                             "input", "raw", "/workspaces/Fall2025-MS3083-Base_Template/data",
                             "/workspaces/Fall2025-MS3083-Base_Template/data/raw",
                             "/workspaces/Fall2025-MS3083-Base_Template/data/clean"
                           )) {
  tokens <- unique(tolower(tokens))
  exts   <- c("csv", "xlsx")
  roots  <- unique(roots)

  # Gather all candidates once
  all_files <- unlist(lapply(roots, function(r) {
    if (!dir.exists(r)) return(character())
    list.files(r, pattern = "(?i)\\.(csv|xlsx)$", recursive = TRUE, full.names = TRUE)
  }))

  if (length(all_files) == 0) {
    stop("No .csv or .xlsx files found under the expected directories.")
  }

  # Score files by how well they match tokens and by simplicity of name
  score_one <- function(path) {
    nm <- tolower(basename(path))
    base <- sub("\\.(csv|xlsx)$", "", nm, ignore.case = TRUE)
    # token hits
    hits <- sum(sapply(tokens, function(tk) grepl(tk, base, fixed = TRUE)))
    # bonus for exact/simple bases
    exact_bonus <- if (any(base %in% tokens)) 2 else 0
    clean_bonus <- if (grepl("(^|[^a-z])clean($|[^a-z])", base)) 1 else 0
    # small penalty for very long names (less likely to be the canonical table)
    length_pen  <- nchar(base) / 100
    hits + exact_bonus + clean_bonus - length_pen
  }

  # Filter to candidates that contain at least one token
  cand <- all_files[sapply(all_files, function(p) {
    base <- tolower(sub("\\.(csv|xlsx)$", "", basename(p)))
    any(sapply(tokens, function(tk) grepl(tk, base, fixed = TRUE)))
  })]

  if (length(cand) == 0) {
    stop(sprintf(
      "Could not find a file containing any of these tokens: %s\nSearched under:\n%s",
      paste(tokens, collapse = ", "),
      paste0(" - ", roots, collapse = "\n")
    ), call. = FALSE)
  }

  # Rank by score (desc) and prefer CSV over XLSX when tied (faster, fewer deps)
  scores <- sapply(cand, score_one)
  ord <- order(scores, decreasing = TRUE)
  cand <- cand[ord]
  # prefer csv on top if scores equal
  top_score <- scores[ord][1]
  tops <- cand[abs(scores[ord] - top_score) < 1e-9]
  csv_first <- c(tops[grepl("(?i)\\.csv$", tops)], tops[grepl("(?i)\\.xlsx$", tops)])
  best <- csv_first[1]
  best
}

# ---- Loaders ---------------------------------------------------------------
load_table <- function(path) {
  ext <- tolower(tools::file_ext(path))
  if (ext == "csv") {
    read.csv(path, stringsAsFactors = FALSE, check.names = FALSE)
  } else if (ext == "xlsx") {
    if (!requireNamespace("readxl", quietly = TRUE)) {
      stop("readxl package is required to read .xlsx files. Please install it (install.packages('readxl')) or provide a CSV.", call. = FALSE)
    }
    readxl::read_xlsx(path) |> as.data.frame(check.names = FALSE)
  } else {
    stop(sprintf("Unsupported file type: %s", ext), call. = FALSE)
  }
}

# Standardize a likely customer ID column to "CustomerID"
standardize_customer_id <- function(df) {
  nms <- names(df)
  if ("CustomerID" %in% nms) return(df)
  candidates <- c(
    "Customer_ID", "Customer.Id", "customer_id", "customerId",
    "Customer Key", "CustID", "Cust_ID", "cust_id", "custid"
  )
  hit <- candidates[candidates %in% nms]
  if (length(hit) == 0) {
    rx <- grep("(?i)customer.*id", nms, perl = TRUE, value = TRUE)
    if (length(rx) == 0) {
      stop("No CustomerID-like column found. Expected 'CustomerID' or a close variant.", call. = FALSE)
    }
    names(df)[match(rx[1], names(df))] <- "CustomerID"
  } else {
    names(df)[match(hit[1], names(df))] <- "CustomerID"
  }
  df
}

# ---- Locate & Load ---------------------------------------------------------
say("Working dir: ", getwd())

customers_path <- find_data_file(tokens = c("customers", "customer"))
orders_path    <- find_data_file(tokens = c("orders", "order"))

say("Customers file -> ", customers_path)
say("Orders file    -> ", orders_path)

customers <- load_table(customers_path)
orders    <- load_table(orders_path)

# Normalize key column name and type
customers <- standardize_customer_id(customers)
orders    <- standardize_customer_id(orders)

customers$CustomerID <- as.character(customers$CustomerID)
orders$CustomerID    <- as.character(orders$CustomerID)

# ---- Join Customers with Orders -------------------------------------------
customer_orders <- customers %>%
  dplyr::left_join(orders, by = "CustomerID")

# ---- Output Summary --------------------------------------------------------
cat("========== CUSTOMER ORDERS ==========\n")
cat("Total rows:", nrow(customer_orders), "\n")
cat("Columns:", ncol(customer_orders), "\n")
cat("Preview:\n")
print(utils::head(customer_orders, 5))

```

```r
# ==========================================
# Task 7.1: Clean Text Data (robust)
# ==========================================

suppressPackageStartupMessages({
  library(dplyr)
  library(stringr)
  library(rlang)
})

find_col <- function(df, candidates) {
  nms <- names(df)
  # exact (case-insensitive)
  for (cand in candidates) {
    hit <- nms[tolower(nms) == tolower(cand)]
    if (length(hit) > 0) return(hit[1])
  }
  # punctuation-insensitive
  norm <- function(x) gsub("[^a-z0-9]", "", tolower(x))
  nms_norm <- norm(nms)
  for (cn in norm(candidates)) {
    idx <- match(cn, nms_norm)
    if (!is.na(idx)) return(nms[idx])
  }
  # heuristic partial
  hits <- grep(paste(candidates, collapse="|"), nms, ignore.case = TRUE, value = TRUE)
  if (length(hits) > 0) return(hits[1])
  NA_character_
}

region_candidates   <- c("Region","region","Region_Name","region_name")
category_candidates <- c("Product_Category","Product.Category","Category",
                         "product_category","Prod_Category","ProductCategory")

src_region <- find_col(sales_enhanced, region_candidates)
src_cat    <- find_col(sales_enhanced, category_candidates)

if (is.na(src_region) && is.na(src_cat)) {
  stop("Neither a Region nor a Product Category column was found. Columns available are: ",
       paste(names(sales_enhanced), collapse = ", "))
}

if (!is.na(src_region)) {
  cat("Using '", src_region, "' as Region source.\n", sep = "")
} else {
  cat("No Region column found -> creating region_clean = NA.\n")
}
if (!is.na(src_cat)) {
  cat("Using '", src_cat, "' as Product Category source.\n", sep = "")
} else {
  cat("No Product Category column found -> creating category_clean = NA.\n")
}

sales_enhanced <- sales_enhanced %>%
  {
    df <- .
    if (!is.na(src_region)) {
      df <- df %>% mutate(region_clean = str_to_title(str_trim(!!sym(src_region))))
    } else {
      df <- df %>% mutate(region_clean = NA_character_)
    }
    if (!is.na(src_cat)) {
      df <- df %>% mutate(category_clean = str_to_title(str_trim(!!sym(src_cat))))
    } else {
      df <- df %>% mutate(category_clean = NA_character_)
    }
    df
  }

cat("========== CLEANED TEXT DATA ==========\n")
head(sales_enhanced %>% select(
  any_of(src_region), region_clean,
  any_of(src_cat),    category_clean
), 5)
```

```r
# ==========================================
# Task 7.2: Parse Dates and Extract Components (robust)
# ==========================================

suppressPackageStartupMessages({
  library(dplyr)
  library(lubridate)
  library(rlang)
})

# Helper to find a column by common name variants
find_col <- function(df, candidates) {
  nms <- names(df)
  # exact (case-insensitive)
  hit <- nms[tolower(nms) %in% tolower(candidates)]
  if (length(hit) > 0) return(hit[1])
  # punctuation-insensitive
  norm <- function(x) gsub("[^a-z0-9]", "", tolower(x))
  nms_norm <- norm(nms); cand_norm <- norm(candidates)
  for (cn in cand_norm) {
    idx <- match(cn, nms_norm)
    if (!is.na(idx)) return(nms[idx])
  }
  # heuristic: any column containing "date"
  hits <- nms[grepl("date", nms, ignore.case = TRUE)]
  if (length(hits) > 0) return(hits[1])
  NA_character_
}

# Common date column names in course datasets
date_candidates <- c(
  "Sale_Date","Sale.Date","sale_date",
  "Purchase_Date","Purchase.Date","purchase_date",
  "Order_Date","Order.Date","order_date",
  "Date","date"
)

src_date <- find_col(sales_enhanced, date_candidates)

if (is.na(src_date)) {
  stop("No date-like column found. Available columns: ",
       paste(names(sales_enhanced), collapse = ", "))
} else {
  cat("Using '", src_date, "' as the source date column.\n", sep = "")
}

sales_enhanced <- sales_enhanced %>%
  mutate(
    # Parse date robustly from multiple possible types
    date_parsed = {
      v <- .[[src_date]]
      if (inherits(v, "Date")) {
        v
      } else if (inherits(v, "POSIXt")) {
        as.Date(v)
      } else if (is.numeric(v)) {
        # Treat as Excel serial (works for most .xlsx exports)
        as.Date(v, origin = "1899-12-30")
      } else {
        suppressWarnings(parse_date_time(as.character(v),
          orders = c("ymd","mdy","dmy","Ymd HMS","mdy HMS","dmy HMS")))
      }
    },
    sale_month   = month(date_parsed, label = TRUE, abbr = FALSE),
    sale_weekday = wday(date_parsed, label = TRUE, abbr = FALSE)
  )

cat("========== DATE COMPONENTS ==========\n")
head(sales_enhanced %>%
       select(any_of(c("Sale_Date","Purchase_Date","Order_Date","Date", src_date)),
              date_parsed, sale_month, sale_weekday), 5)

```

```r
# ==========================================
# Task 8.1: Create Performance Categories (robust)
# ==========================================

suppressPackageStartupMessages({
  library(dplyr)
  library(rlang)
})

# Helper to find a column by common variants (case & punctuation insensitive)
find_col <- function(df, candidates) {
  nms <- names(df)
  # exact case-insensitive
  hit <- nms[tolower(nms) %in% tolower(candidates)]
  if (length(hit) > 0) return(hit[1])
  # punctuation-insensitive
  norm <- function(x) gsub("[^a-z0-9]", "", tolower(x))
  nms_norm <- norm(nms); cand_norm <- norm(candidates)
  for (cn in cand_norm) {
    idx <- match(cn, nms_norm)
    if (!is.na(idx)) return(nms[idx])
  }
  # heuristic: contains "revenue" or "sales"
  hits <- nms[grepl("revenue|sales", nms, ignore.case = TRUE)]
  if (length(hits) > 0) return(hits[1])
  NA_character_
}

revenue_candidates <- c(
  "Revenue","Total_Revenue","Net_Revenue",
  "Sales","Sales_Amount","Sale_Amount","Total_Sales",
  "Amount"
)

src_rev <- find_col(sales_enhanced, revenue_candidates)

if (is.na(src_rev)) {
  stop("No revenue-like column found. Available columns: ",
       paste(names(sales_enhanced), collapse = ", "))
} else {
  cat("Using '", src_rev, "' as the revenue source column.\n", sep = "")
}

sales_enhanced <- sales_enhanced %>%
  mutate(
    # Coerce revenue to numeric (handles "$", "," and character types)
    revenue_num = {
      v <- .data[[src_rev]]
      if (is.numeric(v)) v else suppressWarnings(as.numeric(gsub("[,$]", "", as.character(v))))
    },
    performance_tier = case_when(
      revenue_num > 25000 ~ "High",
      revenue_num > 15000 ~ "Medium",
      !is.na(revenue_num) ~ "Low",
      TRUE ~ NA_character_
    )
  )

cat("========== PERFORMANCE TIERS ==========\n")
print(table(sales_enhanced$performance_tier, useNA = "ifany"))

```

```r
# ==========================================
# Task 8.2: Calculate Business KPIs (robust)
# ==========================================

suppressPackageStartupMessages({
  library(dplyr)
  library(rlang)
})

# Helper to find likely revenue/sales column
find_col <- function(df, candidates) {
  nms <- names(df)
  hit <- nms[tolower(nms) %in% tolower(candidates)]
  if (length(hit) > 0) return(hit[1])
  norm <- function(x) gsub("[^a-z0-9]", "", tolower(x))
  nms_norm <- norm(nms); cand_norm <- norm(candidates)
  for (cn in cand_norm) {
    idx <- match(cn, nms_norm)
    if (!is.na(idx)) return(nms[idx])
  }
  hits <- nms[grepl("revenue|sales", nms, ignore.case = TRUE)]
  if (length(hits) > 0) return(hits[1])
  NA_character_
}

rev_candidates <- c("Revenue","Sales","Sales_Amount","Total_Sales","Total_Revenue","Sale_Amount")
src_rev <- find_col(sales_enhanced, rev_candidates)

if (is.na(src_rev)) {
  stop("No revenue or sales column found. Columns available: ",
       paste(names(sales_enhanced), collapse = ", "))
} else {
  cat("Using '", src_rev, "' as the revenue source column.\n", sep = "")
}

business_kpis <- sales_enhanced %>%
  mutate(
    rev_num = {
      v <- .[[src_rev]]
      if (is.numeric(v)) v else suppressWarnings(as.numeric(gsub("[,$]", "", as.character(v))))
    }
  ) %>%
  summarize(
    total_revenue        = sum(rev_num, na.rm = TRUE),
    total_transactions   = n(),
    avg_transaction_value = mean(rev_num, na.rm = TRUE),
    high_value_pct        = mean(high_value == "Yes", na.rm = TRUE) * 100
  )

cat("========== BUSINESS KPIs ==========\n")
print(business_kpis)
```
